```python

def reciprocal_rank_fusion(rankings, k=60):

    fused_scores = {}

    for ranking in rankings:

        for rank, product_id in enumerate(ranking, start=1):

            if product_id not in fused_scores:
                fused_scores[product_id] = 0.0

            fused_scores[product_id] += 1 / (k + rank)

    return sorted(
        fused_scores.items(),
        key=lambda x: x[1],
        reverse=True
    )

print("RRF function ready!")
```

```python

def hybrid_search(query, top_k=5):

    # 1. Semantic retrieval
    query_embedding = embedding_model.encode(query).tolist()

    vector_results = collection.query(
        query_embeddings=[query_embedding],
        n_results=min(top_k, collection.count())
    )

    vector_ranking = vector_results["ids"][0]

    # 2. Keyword retrieval using BM25
    tokenized_query = tokenize(query)

    bm25_scores = bm25.get_scores(tokenized_query)

    bm25_indices = np.argsort(bm25_scores)[::-1][:top_k]

    bm25_ranking = [
        str(df.iloc[idx]["product_id"])
        for idx in bm25_indices
        if bm25_scores[idx] > 0
    ]

    # 3. Combine rankings using RRF
    fused_results = reciprocal_rank_fusion(
        [vector_ranking, bm25_ranking]
    )

    # 4. Retrieve complete product information
    product_lookup = df.set_index("product_id")

    final_results = []

    for product_id, score in fused_results[:top_k]:

        product = product_lookup.loc[product_id]

        final_results.append({
            "product_id": product_id,
            "name": product["name"],
            "category": product["category"],
            "brand": product["brand"],
            "price": int(product["price"]),
            "description": product["description"],
            "specifications": product["specifications"],
            "warranty": product["warranty"],
            "availability": product["availability"],
            "rrf_score": float(score)
        })

     
    # 5. Filter products by requested category
    import re

    query_lower = query.lower()

    
    category_keywords = {
        "headphones": [
            "headphone",
            "headphones",
            "earbud",
            "earbuds",
            "headset"
        ],
        "smartphone": [
            "smartphone",
            "phone",
            "mobile"
        ],
        "laptop": [
            "laptop",
            "notebook",
            "computer",
            "computers",
            "pc"
        ],
        "tablet": [
            "tablet",
            "ipad"
        ]
    }


    for category, keywords in category_keywords.items():

        if any(
            re.search(r"\b" + re.escape(word) + r"\b", query_lower)
            for word in keywords
        ):
            final_results = [
                product for product in final_results
                if product["category"].lower() == category
            ]
            break

    return final_results
```

```python

import time
from google.genai import errors

def generate_with_retry(prompt, max_retries=4):

    for attempt in range(max_retries):
        try:
            response = gemini_client.models.generate_content(
                model=GEMINI_MODEL,
                contents=prompt,
                config=types.GenerateContentConfig(
                    max_output_tokens=500,
                    temperature=0.4
                )
            )

            return response.text

        except errors.ServerError as e:
            if attempt == max_retries - 1:
                print("Gemini is still unavailable after retries.")
                raise

            wait_time = 2 ** (attempt + 1)

            print(f"Gemini 503 error. Retrying in {wait_time} seconds...")
            time.sleep(wait_time)
```

```python

import re

def filter_products_by_budget(query, products):

    query_lower = query.lower().replace(",", "")

    budget_patterns = [
        r'under\s*₹?\s*(\d+)',
        r'below\s*₹?\s*(\d+)',
        r'within\s*₹?\s*(\d+)',
        r'budget(?:\s+is)?(?:\s+now)?\s*₹?\s*(\d+)',
        r'budget\s+of\s*₹?\s*(\d+)',
        r'₹\s*(\d+)\s*(?:budget|inr)?'
    ]

    budget = None

    for pattern in budget_patterns:
        match = re.search(pattern, query_lower)
        if match:
            budget = int(match.group(1))
            break

    if budget is None:
        return products

    filtered_products = [
        p for p in products
        if float(p.get("price", 0)) <= budget
    ]

    return filtered_products
```

```python

def precision_at_k(retrieved, relevant, k=3):
    retrieved_top_k = retrieved[:k]

    if k == 0:
        return 0.0

    relevant_count = sum(
        1 for product_id in retrieved_top_k
        if product_id in relevant
    )

    return relevant_count / k


def recall_at_k(retrieved, relevant, k=3):
    if not relevant:
        return 0.0

    retrieved_top_k = retrieved[:k]

    relevant_count = sum(
        1 for product_id in retrieved_top_k
        if product_id in relevant
    )

    return relevant_count / len(relevant)


def reciprocal_rank(retrieved, relevant):
    for rank, product_id in enumerate(retrieved, start=1):
        if product_id in relevant:
            return 1 / rank

    return 0.0


def evaluate_retrieval(evaluation_queries, k=3):

    evaluation_results = []

    for item in evaluation_queries:

        query = item["query"]
        relevant = set(item["expected_products"])

        # Retrieve products using the actual hybrid search
        products = hybrid_search(query, top_k=7)

        retrieved_ids = [
            p["product_id"] for p in products
        ]

        precision = precision_at_k(
            retrieved_ids, relevant, k
        )

        recall = recall_at_k(
            retrieved_ids, relevant, k
        )

        mrr = reciprocal_rank(
            retrieved_ids, relevant
        )

        evaluation_results.append({
            "Query": query,
            "Retrieved Products": retrieved_ids[:k],
            "Relevant Products": list(relevant),
            "Precision@K": round(precision, 3),
            "Recall@K": round(recall, 3),
            "Reciprocal Rank": round(mrr, 3)
        })

    return pd.DataFrame(evaluation_results)
```

```python

def get_vector_ids(query, top_k=7):
    query_embedding = embedding_model.encode(query).tolist()

    results = collection.query(
        query_embeddings=[query_embedding],
        n_results=min(top_k, collection.count())
    )

    return results["ids"][0]


def get_bm25_ids(query, top_k=7):
    scores = bm25.get_scores(tokenize(query))
    indices = np.argsort(scores)[::-1][:top_k]

    return [
        str(df.iloc[idx]["product_id"])
        for idx in indices
        if scores[idx] > 0
    ]


def get_hybrid_ids(query, top_k=7):
    results = hybrid_search(query, top_k=top_k)
    return [p["product_id"] for p in results]


def evaluate_method(retrieval_function, queries, k=3):
    scores = []

    for item in queries:
        retrieved = retrieval_function(item["query"], 7)
        relevant = set(item["expected_products"])

        scores.append({
            "Precision@3": precision_at_k(retrieved, relevant, k),
            "Recall@3": recall_at_k(retrieved, relevant, k),
            "MRR": reciprocal_rank(retrieved, relevant)
        })

    return pd.DataFrame(scores).mean().round(3)


comparison = pd.DataFrame({
    "Vector Search": evaluate_method(
        get_vector_ids, evaluation_queries
    ),
    "BM25": evaluate_method(
        get_bm25_ids, evaluation_queries
    ),
    "Hybrid Search": evaluate_method(
        get_hybrid_ids, evaluation_queries
    )
}).T

display(comparison)
```

```python

def calculate_query_metrics(retrieved_ids, expected_ids, k=3):

    retrieved = retrieved_ids[:k]
    expected = set(expected_ids)

    relevant_count = sum(
        1 for pid in retrieved if pid in expected
    )

    precision = relevant_count / k
    recall = (
        relevant_count / len(expected)
        if expected else 0
    )

    reciprocal_rank = 0

    for rank, pid in enumerate(retrieved, start=1):
        if pid in expected:
            reciprocal_rank = 1 / rank
            break

    return precision, recall, reciprocal_rank


query_results = []

for item in combined_evaluation_queries:

    query = item["query"]
    expected = set(item["expected_products"])

    hybrid_ids = get_hybrid_ids(query, top_k=3)
    reranked_ids = get_reranked_ids(query, top_k=3)

    h_p, h_r, h_mrr = calculate_query_metrics(
        hybrid_ids, expected
    )

    c_p, c_r, c_mrr = calculate_query_metrics(
        reranked_ids, expected
    )

    query_results.append({
        "Query": query,
        "Hybrid P@3": h_p,
        "Reranked P@3": c_p,
        "Hybrid Recall@3": h_r,
        "Reranked Recall@3": c_r,
        "Hybrid MRR": h_mrr,
        "Reranked MRR": c_mrr
    })

query_metrics = pd.DataFrame(query_results)

display(query_metrics.round(3))
```
